14. Find Missing Number

Question: A list contains distinct numbers from 1 to n, but one number is missing. Find the missing number.

الترجمة: توجد List تحتوي على أرقام مختلفة من 1 إلى n، لكن رقمًا واحدًا مفقود. أوجد الرقم المفقود.

````python
def find(numbers):
    n = len(numbers) + 1
    all_n = (n * (n + 1)) // 2
    sem = sum(numbers)
    hide = all_n - sem
    
    return hide


numbers = [1, 2, 4, 5]

print(f"The Hidden Number is : {find(numbers)}")
````

````python
lis = [2, 3, 5, 6, 14.55]
target = 8

result = set()

for i in range(len(lis)):
    for j in range(len(lis)):
        if i != j and lis[i] + lis[j] == target:
            result.add(frozenset([lis[i], lis[j]]))

print(result)
````

````python
def two_sum(numbers, target):
    seen = {}

    for index, number in enumerate(numbers):
        needed = target - number

        if needed in seen:
            return seen[needed], index

        seen[number] = index

    return None

numbers = [2, 7, 11, 15]
print(two_sum(numbers, 9))  # (0, 1)
````

17. Longest Consecutive Sequence

Question: Given an unsorted list of integers, find the length of the longest sequence of consecutive numbers.

الترجمة: لديك List غير مرتبة من الأعداد الصحيحة، أوجد طول أطول تسلسل من الأرقام المتتالية.

````python
nums = [100, 4, 200, 1, 3, 2]
num_set = set(nums)
longest = 0
longest_sequence = []
for num in num_set:
    
    if num - 1 not in num_set:
        start =  num
        length = 0
        sequence = []
        
    while start in num_set:
        length += 1
        sequence.append(start)
        start += 1 
        
    if len(sequence) > len(longest_sequence):
        longest_sequence = sequence

print("Longest sequence:", longest_sequence)
print("Length:", len(longest_sequence))
print("Sum:", sum(longest_sequence))
````

18. First Non-Repeating Character

Question: Given a string, find the first character that appears only once. Return None if every character repeats.

الترجمة: لديك String، أوجد أول حرف ظهر مرة واحدة فقط. أرجع None إذا كان كل حرف مكررًا.

````python
text = "taha"
def first_non_repeating(text):

    count = {}

    # Step 1: Count every character
    for char in text:
        if char not in count:
            count[char] = 1
        else:
            count[char] += 1

    # Step 2: Find the first character that appeared once
    for char in text:
        if count[char] == 1:
            return char

    # No unique character
    return None
print(first_non_repeating('tata'))
````

19. Group Anagrams

Question: Given a list of words, group words that are anagrams of each other.

الترجمة: لديك List من الكلمات، اجمع الكلمات التي تتكون من نفس الحروف حتى لو اختلف ترتيبها.

````python
def group_anagrams(words):
    group = {}

    for word in words:
        key = "".join(sorted(word))

        if key not in group:
            group[key] = [word]
        else:
            group[key].append(word)

    return list(group.values())


words = ["eat", "tea", "tan", "ate", "nat", "bat"]

result = group_anagrams(words)

print(result)
````

20. Inventory Management

Question: Given a dictionary containing product names and quantities, process a list of sales. Decrease the quantity for each sold product and report products whose quantity becomes zero or negative.

الترجمة: لديك Dictionary يحتوي على أسماء المنتجات وكمياتها، ونفّذ قائمة من عمليات البيع. قلّل الكمية لكل منتج تم بيعه، وأظهر المنتجات التي أصبحت كميتها صفرًا أو أقل.



````python
def process_sales(inventory, sales):
    result = []

    for product in sales:
        inventory[product] -= 1

        if inventory[product] <= 0:
            result.append(product)

    return result


inventory = {
    "apple": 5,
    "banana": 3,
    "milk": 2,
    "bread": 1
}

sales = ["apple", "banana", "apple", "milk", "bread", "banana", "banana"]

result = process_sales(inventory, sales)

print("Products with zero or negative quantity:", result)
````

## 12. Counting Pattern

## 13. Grouping Pattern

```python
groups = {}

    key = some_transformation(item)

    else:
```

```python
students = {
```

```python
groups = {}

    if grade not in groups:
        groups[grade] = [name]
    else:
        groups[grade].append(name)
```

# Python Set — Cheat Sheet

## 1. Creating a Set

```python
numbers = {1, 2, 3, 4}
```

Duplicates are automatically removed:

```python
numbers = {1, 2, 2, 3, 3, 4}
```

Result:

```python
{1, 2, 3, 4}
```

---

## 2. Important: Empty Set

This:

```python
x = {}
```

creates an empty **dictionary**, not a set.

For an empty set:

```python
x = set()
```

---

## 3. Adding Elements

### `add()`

Adds one element.

```python
numbers.add(5)
```

### `update()`

Adds multiple elements.

```python
numbers.update([6, 7, 8])
```

---

## 4. Removing Elements

### `remove()`

```python
numbers.remove(3)
```

If the element does not exist, it raises an error.

### `discard()`

```python
numbers.discard(3)
```

If the element does not exist, nothing happens.

### `pop()`

```python
numbers.pop()
```

Removes an arbitrary element.

### `clear()`

```python
numbers.clear()
```

Removes everything.

---

## 5. Membership Testing

One of the most important uses of a Set:

```python
3 in numbers
```

```python
10 not in numbers
```

Sets are especially useful when you repeatedly need to ask:

```text
"Does this element exist?"
```

---

# Set Operations

Suppose:

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
```

---

## 6. Union

All elements from both sets.

```python
A | B
```

Or:

```python
A.union(B)
```

Result:

```python
{1, 2, 3, 4, 5, 6}
```

---

## 7. Intersection

Elements that exist in both sets.

```python
A & B
```

Or:

```python
A.intersection(B)
```

Result:

```python
{3, 4}
```

---

## 8. Difference

Elements in `A` but not in `B`.

```python
A - B
```

Result:

```python
{1, 2}
```

The reverse:

```python
B - A
```

Result:

```python
{5, 6}
```

---

## 9. Symmetric Difference

Elements that exist in either set, but not in both.

```python
A ^ B
```

Or:

```python
A.symmetric_difference(B)
```

Result:

```python
{1, 2, 5, 6}
```

---
